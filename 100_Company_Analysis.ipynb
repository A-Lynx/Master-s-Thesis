```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- Ielādē datu kopu ---
df = pd.read_csv("100_company_main.csv")

# --- Sustainalytics riska līmeņu sliekšņi ---
SLIEKSNI = [10, 20, 30, 40]

def pievienot_sliekšņu_līnijas(ass):
    """
    Pievieno Sustainalytics riska līmeņu vertikālās atsauces līnijas histogrammā.
    """
    for s in SLIEKSNI:
        ass.axvline(s, linestyle="--", linewidth=1, color="gray", alpha=0.7)
    ass.set_ylabel("Uzņēmumu skaits")

# --- Kopējā ESG rādītāja sadalījuma histogramma ---
esg = df["Total_ESG"].dropna()

videja_vertiba = esg.mean()
mediana = esg.median()

bin_skaits = 15  # piemērots aptuveni 100 novērojumiem

plt.figure(figsize=(8,5))

plt.hist(esg, bins=bin_skaits)

plt.axvline(videja_vertiba,
            linestyle="--",
            linewidth=2,
            color="red",
            label=f"Vidējā vērtība = {videja_vertiba:.2f}")

plt.axvline(mediana,
            linestyle=":",
            linewidth=2,
            color="green",
            label=f"Mediāna = {mediana:.2f}")

pievienot_sliekšņu_līnijas(plt.gca())

plt.title("Kopējā ESG riska rādītāja sadalījums (ar Sustainalytics sliekšņiem)")
plt.xlabel("Kopējais ESG riska rādītājs")

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- Ielādē datu kopu ---
df = pd.read_csv("100_company_main.csv")

# --- Funkcija ESG riska līmeņa noteikšanai pēc Sustainalytics skalas ---
def noteikt_esg_riska_līmeni(vērtība):
    if vērtība < 10:
        return "Nenozīmīgs (0–10)"
    elif vērtība < 20:
        return "Zems (10–20)"
    elif vērtība < 30:
        return "Vidējs (20–30)"
    elif vērtība < 40:
        return "Augsts (30–40)"
    else:
        return "Ļoti augsts (>40)"

# --- Izveido jaunu kolonnu ar riska līmeni ---
df["ESG_riska_līmenis"] = df["Total_ESG"].apply(noteikt_esg_riska_līmeni)

# --- Sakārtots riska līmeņu secīgums ---
līmeņu_secība = [
    "Nenozīmīgs (0–10)",
    "Zems (10–20)",
    "Vidējs (20–30)",
    "Augsts (30–40)",
    "Ļoti augsts (>40)"
]

# --- Aprēķina uzņēmumu skaitu katrā līmenī ---
skaits = df["ESG_riska_līmenis"].value_counts().reindex(līmeņu_secība).fillna(0).astype(int)
procenti = (skaits / skaits.sum() * 100).round(1)

# --- Aprēķina procentuālo sadalījumu ---
procenti = (skaits / skaits.sum() * 100).round(1)

# --- Izveido kopsavilkuma tabulu ---
riska_tabula = pd.DataFrame({
    "Uzņēmumu skaits": skaits,
    "Procenti (%)": procenti
})

print("\nESG riska līmeņu sadalījums:")
print(riska_tabula)

# --- Bar chart vizualizācija ---
plt.figure(figsize=(8,5))

plt.bar(riska_tabula.index, riska_tabula["Uzņēmumu skaits"])

plt.title("Uzņēmumu sadalījums pēc Sustainalytics ESG riska līmeņiem")
plt.xlabel("ESG riska līmenis")
plt.ylabel("Uzņēmumu skaits")

plt.xticks(rotation=25, ha="right")

plt.tight_layout()
plt.show()
```

### Subtask:
Re-execute the code that generates the three ESG sub-component histograms (Environmental, Social, Governance) and save each one as a separate PNG image file (e.g., 'environmental_histogram.png', 'social_histogram.png', 'governance_histogram.png').


**Reasoning**:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# --- Ielādē datu kopu ---
df = pd.read_csv("100_company_main.csv")

# --- ESG apakškomponentu kolonnu saraksts ---
komponentes = ["Environmental", "Social", "Governance"]

# --- Tulkojumu karte virsrakstiem ---
nosaukumu_karte = {
    "Environmental": "Vides risks (E)",
    "Social": "Sociālais risks (S)",
    "Governance": "Pārvaldības risks (G)"
}

# --- Histogrammas parametri ---
bin_skaits = 15  # piemērots aptuveni 100 novērojumiem

# --- Vienotas x ass robežas korektam salīdzinājumam ---
min_vertiba = df[komponentes].min().min()
max_vertiba = df[komponentes].max().max()

# --- Zīmē histogrammas katrai apakškomponentei ---
for kolonna in komponentes:
    dati = df[kolonna].dropna()

    plt.figure(figsize=(8,5))
    plt.hist(dati, bins=bin_skaits)

    plt.title(f"{nosaukumu_karte[kolonna]} sadalījums")
    plt.xlabel("Riska rādītājs")
    plt.ylabel("Uzņēmumu skaits")

    plt.xlim(min_vertiba, max_vertiba)

    plt.tight_layout()
    plt.savefig(f'{kolonna.lower()}_histogram.png') # Save the plot as a PNG file
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

# -----------------------------
# Ielādē datus
# -----------------------------
df = pd.read_csv("100_company_main.csv")

METRIKA_KOL = "DebtToEquity"
METRIKA_NOSAUKUMS = "Parāds pret pašu kapitālu (Debt-to-Equity)"
PDF_FAILS = "Debt_to_Equity_EDA.pdf"

x = pd.to_numeric(df[METRIKA_KOL], errors="coerce").dropna()

# Drošības pārbaude (log iespējams tikai pozitīvām vērtībām)
x = x[x > 0]

# Log transformācija bez +1
x_log = np.log10(x)

# -----------------------------
# Aprakstošā statistika
# -----------------------------
stat = x.describe(percentiles=[0.25, 0.5, 0.75])

stat_df = pd.DataFrame({
    "Rādītājs": [
        "Skaits",
        "Vidējā vērtība",
        "Standartnovirze",
        "Minimums",
        "1. kvartile (25%)",
        "Mediāna (50%)",
        "3. kvartile (75%)",
        "Maksimums",
        "IQR"
    ],
    "Vērtība": [
        stat["count"],
        stat["mean"],
        stat["std"],
        stat["min"],
        stat["25%"],
        stat["50%"],
        stat["75%"],
        stat["max"],
        stat["75%"] - stat["25%"]
    ]
})

stat_df["Vērtība"] = stat_df["Vērtība"].round(2)

display(stat_df)

# -----------------------------
# Vizualizācija
# -----------------------------
fig = plt.figure(figsize=(12, 8))
gs = fig.add_gridspec(2, 2, height_ratios=[2.2, 1.2], hspace=0.4, wspace=0.3)

ax1 = fig.add_subplot(gs[0, 0])
ax2 = fig.add_subplot(gs[0, 1])
ax_tbl = fig.add_subplot(gs[1, :])

# ---- RAW histogramma
mean_raw = x.mean()
median_raw = x.median()

ax1.hist(x, bins=15)
ax1.axvline(mean_raw, linestyle="--", linewidth=2, color="red",
            label=f"Vidējā vērtība = {mean_raw:.2f}")
ax1.axvline(median_raw, linestyle=":", linewidth=2, color="green",
            label=f"Mediāna = {median_raw:.2f}")

ax1.set_title("RAW sadalījums")
ax1.set_xlabel("Debt-to-Equity attiecība")
ax1.set_ylabel("Skaits")
ax1.legend()

# ---- LOG histogramma (bez +1)
mean_log = x_log.mean()
median_log = np.median(x_log)

ax2.hist(x_log, bins=15)
ax2.axvline(mean_log, linestyle="--", linewidth=2, color="red",
            label=f"Vidējā vērtība = {mean_log:.2f}")
ax2.axvline(median_log, linestyle=":", linewidth=2, color="green",
            label=f"Mediāna = {median_log:.2f}")

ax2.set_title("log10 sadalījums")
ax2.set_xlabel("log10(Debt-to-Equity)")
ax2.set_ylabel("Skaits")
ax2.legend()

# ---- Tabula
ax_tbl.axis("off")

ax_tbl.text(
    0.5, 1.15,
    "Aprakstošā statistika",
    ha="center",
    va="bottom",
    fontsize=12,
    fontweight="bold",
    transform=ax_tbl.transAxes
)

table = ax_tbl.table(
    cellText=stat_df.values,
    colLabels=stat_df.columns,
    cellLoc="center",
    loc="center"
)

table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 1.4)

fig.suptitle("Debt-to-Equity (D/E) — Izpētes datu analīze", fontsize=14)
plt.tight_layout()
plt.show()

# Saglabā PDF
with PdfPages(PDF_FAILS) as pdf:
    pdf.savefig(fig, bbox_inches="tight")

print(f"PDF saglabāts: {PDF_FAILS}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

# =========================
# Iestatījumi
# =========================
DATA_FILE = "100_company_main.csv"   # pielāgo, ja citādi
COL = "DebtToEquity"                 # kolonnas nosaukums tavā CSV

OUT_PDF_HIST = "D_E_hist_stat.pdf"
OUT_PDF_BOX  = "D_E_boxplot.pdf"
OUT_PDF_ALL  = "D_E_kopaa.pdf"

BINS_RAW = 15
BINS_LOG = 15

# =========================
# Datu ielāde
# =========================
df = pd.read_csv(DATA_FILE)
x = df[COL].dropna().astype(float)

# PIEŅĒMUMS: Yahoo Finance D/E bieži ir procentos (piem., 135.83 nozīmē 135.83%)
# Tāpēc mēs strādājam ar "x" kā ar procentiem un arī asis marķējam kā (%).
x_pct = x

# Log transformācija: lietojam tikai pozitīvas vērtības, jo log10(<=0) nav definēts
x_pct_pos = x_pct[x_pct > 0]
x_log = np.log10(x_pct_pos)

# =========================
# Statistika (procentos)
# =========================
stats = {
    "Skaits": x_pct.count(),
    "Vidējā vērtība (%)": x_pct.mean(),
    "Standartnovirze (%)": x_pct.std(),
    "Minimums (%)": x_pct.min(),
    "1. kvartile (25%)": x_pct.quantile(0.25),
    "Mediāna (50%)": x_pct.median(),
    "3. kvartile (75%)": x_pct.quantile(0.75),
    "Maksimums (%)": x_pct.max(),
}
stats["IQR (%)"] = stats["3. kvartile (75%)"] - stats["1. kvartile (25%)"]

# =========================
# Palīgfunkcija tabulai
# =========================
def add_stats_table(ax, stats_dict, title):
    ax.axis("off")

    rows = [["Rādītājs", "Vērtība"]]
    for k, v in stats_dict.items():
        if k == "Skaits":
            rows.append([k, f"{int(v)}"])
        else:
            rows.append([k, f"{v:.2f}"])

    table = ax.table(cellText=rows, loc="center", cellLoc="center")
    table.auto_set_font_size(False)
    table.set_fontsize(10)
    table.scale(1, 1.35)

    # header bold
    for (r, c), cell in table.get_celld().items():
        if r == 0:
            cell.set_text_props(fontweight="bold")

    ax.set_title(title, fontsize=12, fontweight="bold", pad=12)

# =========================
# 1) FIGŪRA: histogrammas + statistika (viena lapa)
# =========================
def make_hist_stat_figure():
    fig = plt.figure(figsize=(14, 8))
    gs = fig.add_gridspec(nrows=2, ncols=2, height_ratios=[2.2, 1.2])

    ax_raw = fig.add_subplot(gs[0, 0])
    ax_log = fig.add_subplot(gs[0, 1])
    ax_tbl = fig.add_subplot(gs[1, :])

    # RAW histogramma
    ax_raw.hist(x_pct, bins=BINS_RAW)
    ax_raw.axvline(x_pct.mean(), linestyle="--", linewidth=2, color="red",
                   label=f"Vidējā vērtība = {x_pct.mean():.2f}%")
    ax_raw.axvline(x_pct.median(), linestyle=":", linewidth=2, color="green",
                   label=f"Mediāna = {x_pct.median():.2f}%")
    ax_raw.set_title("RAW sadalījums", fontsize=12, fontweight="bold")
    ax_raw.set_xlabel("Debt-to-Equity (%)")
    ax_raw.set_ylabel("Skaits")
    ax_raw.legend()

    # LOG histogramma (tikai pozitīvās vērtības)
    ax_log.hist(x_log, bins=BINS_LOG)
    ax_log.axvline(x_log.mean(), linestyle="--", linewidth=2, color="red",
                   label=f"Vidējā vērtība = {x_log.mean():.2f}")
    ax_log.axvline(np.median(x_log), linestyle=":", linewidth=2, color="green",
                   label=f"Mediāna = {np.median(x_log):.2f}")
    ax_log.set_title("log10 sadalījums (tikai D/E > 0)", fontsize=12, fontweight="bold")
    ax_log.set_xlabel("log10(Debt-to-Equity (%))")
    ax_log.set_ylabel("Skaits")
    ax_log.legend()

    # Tabula
    add_stats_table(ax_tbl, stats, "Aprakstošā statistika (Debt-to-Equity, %)")

    fig.suptitle("Debt-to-Equity (D/E) — izpētes datu analīze", fontsize=16, fontweight="bold")
    fig.tight_layout(rect=[0, 0, 1, 0.93])
    return fig

# =========================
# 2) FIGŪRA: boxplot (viena lapa)
# =========================
def make_boxplot_figure():
    fig, ax = plt.subplots(figsize=(10, 6))

    ax.boxplot(x_pct, vert=True)

    ax.set_title("Parāds pret pašu kapitālu (Debt-to-Equity) — boxplot", fontsize=14, fontweight="bold")
    ax.set_ylabel("Debt-to-Equity (%)")

    # īsa anotācija ar galvenajām vērtībām
    q1 = stats["1. kvartile (25%)"]
    med = stats["Mediāna (50%)"]
    q3 = stats["3. kvartile (75%)"]
    iqr = stats["IQR (%)"]
    mean = stats["Vidējā vērtība (%)"]

    text = (
        f"Vidējā vērtība: {mean:.2f}%\n"
        f"Mediāna: {med:.2f}%\n"
        f"IQR: {iqr:.2f}%\n"
        f"Q1: {q1:.2f}%\n"
        f"Q3: {q3:.2f}%"
    )
    ax.text(1.15, 0.15, text, transform=ax.transAxes, fontsize=10, va="bottom")

    fig.tight_layout()
    return fig

# =========================
# Izveido figūras
# =========================
fig1 = make_hist_stat_figure()
fig2 = make_boxplot_figure()

# =========================
# 1) Parādi notebookā
# =========================
plt.show()  # parāda fig1 un fig2 (Jupyter parāda abas)

# =========================
# 2) Saglabā atsevišķos PDF
# =========================
fig1.savefig(OUT_PDF_HIST, format="pdf", bbox_inches="tight")
fig2.savefig(OUT_PDF_BOX, format="pdf", bbox_inches="tight")

# =========================
# 3) Saglabā vienā kopējā PDF (2 lapas)
# =========================
with PdfPages(OUT_PDF_ALL) as pdf:
    pdf.savefig(fig1, bbox_inches="tight")
    pdf.savefig(fig2, bbox_inches="tight")

# Aizver figūras (lai nenokrājas atmiņā)
plt.close(fig1)
plt.close(fig2)

print("Saglabāts:")
print(f"- {OUT_PDF_HIST}")
print(f"- {OUT_PDF_BOX}")
print(f"- {OUT_PDF_ALL}")
```

Boxplots ir lielisks veids, kā vizualizēt datu sadalījumu un identificēt ekstrēmās vērtības. Šajā "Parāds pret pašu kapitālu (Debt-to-Equity)" boxplotā redzam sekojošās komponentes:

*   **Centrālā līnija (sarkanā svītra)**: Tā ir **mediāna** (50. percentīle) Debt-to-Equity attiecībai, kas norāda datu kopas vidējo vērtību. Mūsu gadījumā mediāna ir **66.07**, kas nozīmē, ka pusei uzņēmumu Debt-to-Equity attiecība ir zemāka par 66.07, un otrai pusei augstāka.
*   **Kaste (zilais taisnstūris)**: Kaste attēlo **starpkvartīļu diapazonu (IQR)**, kas aptver vidējos 50% datu. Kastes apakšējā mala ir **pirmā kvartīle (Q1)** (25. percentīle) **31.56**, bet augšējā mala ir **trešā kvartīle (Q3)** (75. percentīle) **142.40**. Šis diapazons (IQR = 110.83) mums stāsta par datu izkliedi.
*   **Ūsas (whiskers)**: Ūsas stiepjas no kastes līdz mazākajai un lielākajai vērtībai datu kopā, kas nav atzīmētas kā atsevišķi punkti (ekstrēmās vērtības vai 'outliers'). Tās parasti sasniedz 1.5 reizes IQR attālumā no Q1 un Q3.
*   **Punkti ārpus ūsām**: Šie punkti (ja tādi ir) apzīmē **ekstrēmās vērtības jeb izlecējus (outliers)**, kas atrodas tālāk par 1.5 reizēm IQR no Q1 vai Q3. Mūsu boxplotā ir redzami vairāki atsevišķi punkti virs augšējās ūsas, kas norāda uz uzņēmumiem ar ievērojami augstāku parādu pret pašu kapitālu attiecību salīdzinājumā ar lielāko daļu datu kopas.

Kopumā, boxplots parāda, ka Debt-to-Equity rādītāji ir plaši izkliedēti un ir daži uzņēmumi ar ļoti augstu parādu līmeni, kas varētu signalizēt par lielāku finanšu risku. Mediāna ir mazāka nekā vidējā vērtība (135.83), kas liecina par asimetrisku sadalījumu ar garu labo asti, ko arī apstiprina izlecēji.

```python
# ===============================
# 1️⃣ Import libraries
# ===============================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set(style="white")
plt.rcParams.update({
    "figure.figsize": (12,6),
    "axes.titlesize": 13,
    "axes.labelsize": 11,
    "xtick.labelsize": 9,
    "ytick.labelsize": 9
})

# ===============================
# 2️⃣ Load dataset
# ===============================
df = pd.read_csv("100_company_main.csv")

print("Dataset loaded:", df.shape)

# ===============================
# 3️⃣ Define groups
# ===============================

# Panel A – strictly positive / stable mean
positive_cols = [
    'MarketCap',
    'TotalRevenue',
    'DebtToEquity',
    'CurrentRatio',
    'QuickRatio'
]

# Panel B – may contain negative values
sensitive_cols = [
    'NetIncome',
    'ReturnOnEquity'
]

labels = {
    'MarketCap': "Tirgus kapitalizācija\n(Market Cap)",
    'TotalRevenue': "Ieņēmumi\n(Total Revenue)",
    'DebtToEquity': "Parāda/kapitāla attiecība\n(Debt-to-Equity)",
    'CurrentRatio': "Kopējās likviditātes koef.\n(Current Ratio)",
    'QuickRatio': "Ātrās likviditātes koef.\n(Quick Ratio)",
    'NetIncome': "Tīrā peļņa\n(Net Income)",
    'ReturnOnEquity': "Pašu kapitāla atdeve\n(Return on Equity)"
}

# ===============================
# 4️⃣ Calculate CV
# ===============================
cv_positive = df[positive_cols].std() / df[positive_cols].mean()
cv_sensitive = df[sensitive_cols].std() / df[sensitive_cols].mean()

print("\nCV – Positive Metrics:")
print(cv_positive.round(4))

print("\nCV – Sensitive Metrics:")
print(cv_sensitive.round(4))

# ===============================
# 5️⃣ Plot
# ===============================
fig, axes = plt.subplots(1, 2, figsize=(14,5))

# Panel A
axes[0].bar([labels[c] for c in cv_positive.index], cv_positive.values)
axes[0].set_title("A) Relatīvā izkliede (CV) – pozitīvie rādītāji")
axes[0].set_ylabel("CV = σ / μ")
axes[0].tick_params(axis='x', rotation=45)

# Panel B
axes[1].bar([labels[c] for c in cv_sensitive.index], cv_sensitive.values)
axes[1].set_title("B) CV – interpretēt piesardzīgi\n(negatīvas vērtības / mean jutīgums)")
axes[1].tick_params(axis='x', rotation=45)

sns.despine()
plt.tight_layout()

# ===============================
# 6️⃣ Save PDF
# ===============================
output_file = "cv_finansu_raditaji_paplasinata_versija.pdf"
plt.savefig(output_file, format="pdf", bbox_inches="tight")

print(f"\nFigure saved as: {output_file}")

plt.show()
```
