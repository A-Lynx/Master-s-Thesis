````python
# --- 1) Definējam valstu sadalījumu pa reģioniem (tavs 5 reģionu modelis) ---

regions = {
    "North America": [
        "United States",
        "Canada"
    ],
    "Europe": [
        "United Kingdom",
        "France",
        "Germany",
        "Switzerland",
        "Sweden",
        "Netherlands",
        "Italy",
        "Spain",
        "Ireland",
        "Denmark",
        "Finland",
        "Belgium",
        "Norway",
        "Portugal",
        "Austria",
        "Hungary",
        "Czech Republic",
        "Poland",
        "Greece"
    ],
    "Asia-Pacific": [
        "China",
        "Japan",
        "India",
        "Taiwan",
        "Korea (South)",
        "Australia",
        "Hong Kong",
        "Singapore",
        "Malaysia",
        "Thailand",
        "Indonesia",
        "Philippines",
        "New Zealand"
    ],
    "Africa/Middle East": [
        "Saudi Arabia",
        "South Africa",
        "United Arab Emirates",
        "Qatar",
        "Kuwait",
        "Egypt",
        "Israel",
        "Turkey"
    ],
    "Other": [
        "Brazil",
        "Mexico",
        "Chile",
        "Peru",
        "Colombia",
        "Russian Federation"
    ]
}

# --- 2) Uztaisām country -> region mapping dictionary ---
country_to_region = {
    country: region
    for region, countries in regions.items()
    for country in countries
}

# --- 3) Pievienojam Region kolonu ---
# (drošībai: apgriežam atstarpes Location laukā, ja tādas ir)
acwi_equity['Location'] = acwi_equity['Location'].astype(str).str.strip()

acwi_equity['Region'] = acwi_equity['Location'].map(country_to_region)

# --- 4) Validācija: vai ir valstis bez mapping? ---
unmapped_countries = (
    acwi_equity.loc[acwi_equity['Region'].isna(), 'Location']
    .drop_duplicates()
    .sort_values()
)

print(f"Unmapped valstu skaits: {len(unmapped_countries)}")
if len(unmapped_countries) > 0:
    print("Valstis bez reģiona mapping:")
    print(unmapped_countries.to_list())

# --- 5) Reģionu sadalījums un proporcijas (uzņēmumu skaits) ---
region_counts = acwi_equity['Region'].value_counts(dropna=False)
region_shares = (region_counts / region_counts.sum() * 100).round(2)

print("\nReģionu sadalījums (skaits):")
print(region_counts)

print("\nReģionu proporcijas (%):")
print(region_shares)
````

````python
import pandas as pd

df = pd.read_csv('acwi_552_with_yahoo_names.csv')
display(df.head())
````

**Reasoning**:
To begin building the 100-company sample, I will first extract all companies from the specified non-North American regions ('Europe', 'Asia-Pacific', 'Other', and 'Africa/Middle East') from the `df` DataFrame. This forms the initial part of our sample and helps determine how many more companies are needed from North America.



````python
non_na_regions = ['Europe', 'Asia-Pacific', 'Other', 'Africa/Middle East']
sample_non_na = df[df['Region'].isin(non_na_regions)].copy()

print(f"Companies from non-North American regions: {len(sample_non_na)}")
display(sample_non_na.head())
````

**Reasoning**:
Now that the non-North American companies are selected, I need to calculate how many more companies are required to reach the 100-company target. Then, I will filter the DataFrame to get only North American companies and proportionally sample from 'United States' and 'Canada' to fulfill the remaining count, ensuring reproducibility with a random state.



````python
target_sample_size = 100
remaining_needed = target_sample_size - len(sample_non_na)

# Filter North American companies
df_north_america = df[df['Region'] == 'North America'].copy()

# Split into United States and Canada
df_us = df_north_america[df_north_america['Country'] == 'United States']
df_canada = df_north_america[df_north_america['Country'] == 'Canada']

# Calculate proportional sample sizes for US and Canada
us_sample_size = round(remaining_needed * 0.85)
canada_sample_size = remaining_needed - us_sample_size # Ensure total sums up to remaining_needed

# Ensure sample sizes don't exceed available companies
us_sample_size = min(us_sample_size, len(df_us))
canada_sample_size = min(canada_sample_size, len(df_canada))

# Sample from United States and Canada with a random state for reproducibility
sampled_us = df_us.sample(n=us_sample_size, random_state=42)
sampled_canada = df_canada.sample(n=canada_sample_size, random_state=42)

# Combine all parts of the sample
final_sample = pd.concat([sample_non_na, sampled_us, sampled_canada])

print(f"Final sample size: {len(final_sample)}")
print("\nRegional distribution of the final sample:")
print(final_sample['Region'].value_counts())

display(final_sample.head())
````

## Load 552 Companies Data

### Subtask:
Load the `acwi_552_with_yahoo_names.csv` file into a pandas DataFrame. This DataFrame contains the Tickers for which we need to fetch exchange information.


**Reasoning**:
To load the specified CSV file into a pandas DataFrame, I will use the `pd.read_csv()` function and then display the first few rows to confirm successful loading as instructed.



````python
import pandas as pd

df = pd.read_csv('acwi_552_with_yahoo_names.csv')
display(df.head())
````

**Reasoning**:
To fetch the exchange information for each unique ticker from the `df` DataFrame, I will use the `yfinance` library. I will iterate through each ticker, attempt to retrieve its information, extract the 'exchange' field, and handle any errors or missing data. Finally, I will compile this information into a new DataFrame.



````python
import yfinance as yf
import pandas as pd

# Initialize an empty dictionary to store the fetched exchange data
exchange_info = {}

# Get a list of unique 'Ticker' symbols from the df DataFrame
unique_tickers = df['Ticker'].dropna().unique()

print(f"Fetching exchange information for {len(unique_tickers)} unique Ticker symbols...")

# Loop through each 'Ticker' symbol
for ticker_symbol in unique_tickers:
    try:
        # Create a yfinance.Ticker object
        ticker = yf.Ticker(ticker_symbol)
        # Retrieve the info attribute
        info = ticker.info

        # Check if the 'exchange' field is present
        if 'exchange' in info and info['exchange'] is not None:
            exchange_info[ticker_symbol] = info['exchange']
        else:
            exchange_info[ticker_symbol] = None # If exchange is not found or is None
            # print(f"Warning: Exchange information not found for {ticker_symbol}.")

    except Exception as e:
        exchange_info[ticker_symbol] = None # Store None if any error occurs during fetching
        # print(f"Error fetching information for {ticker_symbol}: {e}")

# Convert the exchange_info dictionary into a pandas DataFrame
yahoo_finance_exchanges_df = pd.DataFrame(list(exchange_info.items()), columns=['Ticker', 'Yahoo_Finance_Exchange'])

# Display the head of the newly created DataFrame
print("\nFetched Yahoo Finance Exchange information (head):")
display(yahoo_finance_exchanges_df.head())

# Display summary of fetched exchanges
missing_exchanges_count = yahoo_finance_exchanges_df['Yahoo_Finance_Exchange'].isnull().sum()
print(f"\nTotal tickers with fetched exchange info: {len(yahoo_finance_exchanges_df) - missing_exchanges_count}")
print(f"Total tickers with missing exchange info: {missing_exchanges_count}")
````

**Reasoning**:
To incorporate the newly fetched Yahoo Finance exchange information into the main dataset, I will merge the `yahoo_finance_exchanges_df` with the `df` DataFrame on the 'Ticker' column using a left merge. This ensures that `df` is updated with the exchange data while retaining all its original entries.



````python
df = pd.merge(df, yahoo_finance_exchanges_df, on='Ticker', how='left')

print("Updated df DataFrame with Yahoo Finance Exchange information (head):")
display(df.head())

missing_merged_exchanges_count = df['Yahoo_Finance_Exchange'].isnull().sum()
print(f"\nTotal tickers with missing Yahoo Finance Exchange info after merge: {missing_merged_exchanges_count}")
````
